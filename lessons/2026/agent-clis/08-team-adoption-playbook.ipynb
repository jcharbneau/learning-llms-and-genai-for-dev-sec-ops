## Code Walkthrough
The next two code cells are intentionally split:

```python
# Deterministic Demo
def rollout_plan(maturity: str):
    base = ["define guardrails", "pilot one team", "measure quality and cycle time"]
    if maturity == "advanced":
        base += ["enable multi-agent workflows", "weekly governance review"]
    else:
        base += ["reviewer checklist", "safe-default training"]
    return base

plan = rollout_plan("intermediate")
for i, step in enumerate(plan, 1):
    print(i, step)

assert len(plan) >= 5
```

```python
# Live Demo
import shutil
import subprocess

prompt = "Create a phased team rollout plan for coding agents with governance checkpoints and measurable success criteria."
cmds = [
    ["openclaw", "agent", "--local", "--to", "+15555550123", "--message", prompt, "--timeout", "70"],
    ["codex", "exec", prompt],
    ["claude", "-p", prompt],
    ["opencode", "run", prompt],
]

for cmd in cmds:
    if shutil.which(cmd[0]) is None:
        continue
    p = subprocess.run(cmd, capture_output=True, text=True, timeout=70)
    if p.returncode == 0:
        out = (p.stdout or p.stderr).strip()
        print(out[:1200])
        assert "phase" in out.lower() or "rollout" in out.lower()
        break
else:
    print("Skipping live rollout generation: no compatible non-interactive CLI available.")
```
