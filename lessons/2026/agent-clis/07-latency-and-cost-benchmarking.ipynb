## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: executes a real provider/CLI flow with explicit graceful-skip behavior.


```python
# Deterministic Demo
bench = [
    {"tool": "codex", "lat_s": 9.2, "cost": 0.14, "success": 0.92},
    {"tool": "claude", "lat_s": 8.5, "cost": 0.16, "success": 0.94},
    {"tool": "opencode", "lat_s": 7.8, "cost": 0.10, "success": 0.86},
]

for row in bench:
    row["score"] = round((row["success"] * 2) - (row["lat_s"] / 20) - row["cost"], 3)

ranked = sorted(bench, key=lambda x: x["score"], reverse=True)
print(ranked)
assert len(ranked) == 3
```

```python
# Live Demo
import shutil
import subprocess
import time

prompt = "In one sentence, define deterministic benchmarking for AI developer tools."
cmds = [["codex", "exec", prompt], ["claude", "-p", prompt], ["opencode", "run", prompt]]

for cmd in cmds:
    if shutil.which(cmd[0]) is None:
        continue
    t0 = time.perf_counter()
    p = subprocess.run(cmd, capture_output=True, text=True, timeout=70)
    dt = round(time.perf_counter() - t0, 2)
    print({"tool": cmd[0], "rc": p.returncode, "latency_s": dt})
    if p.returncode == 0:
        print((p.stdout or p.stderr).strip()[:1000])
    break
else:
    print("Skipping live benchmark run: no compatible non-interactive CLI available.")
```
