# Agent CLIs Playbook: OpenCode, Codex, Claude Code (2026)

Goal: compare agent coding CLIs in a safe, reproducible way for Dev/Sec/Ops teams.


```python
import shutil
import subprocess

TOOLS = {
    "opencode": ["opencode", "--help"],
    "codex": ["codex", "--help"],
    "claude": ["claude", "--help"],
}

for name, cmd in TOOLS.items():
    available = shutil.which(cmd[0]) is not None
    print(f"{name}: available={available}")
```

## Capability smoke checks
Run each CLI help command when available; skip otherwise.


```python
results = {}
for name, cmd in TOOLS.items():
    if shutil.which(cmd[0]) is None:
        results[name] = "SKIP (not installed)"
        continue
    proc = subprocess.run(cmd, capture_output=True, text=True, timeout=20)
    ok = proc.returncode == 0
    results[name] = "OK" if ok else f"FAIL ({proc.returncode})"
    snippet = (proc.stdout or proc.stderr)[:300]
    print(f"[{name}] status={results[name]}")
    print(snippet)

print(results)
assert isinstance(results, dict)
```

## Suggested evaluation matrix
- Sandbox controls and permission prompts
- Test execution fidelity and deterministic reruns
- Diff quality and commit hygiene
- Cost, latency, and model/provider flexibility
