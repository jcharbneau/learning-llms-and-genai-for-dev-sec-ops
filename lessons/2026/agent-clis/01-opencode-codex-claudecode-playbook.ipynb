# 01 Capability Benchmarking Across Agent CLIs (Agent-CLIs, 2026)

## What This Lesson Is
Compare CLI agent capabilities with measurable criteria and reproducible benchmark prompts.

## Scientific Lens
- Concept: Tool capability benchmarking
- Measure: Task success rate, latency, and output quality rubric score
- Validity Limit: CLI behavior can change across versions; benchmarks require regular refresh.


## How It Works
1. Define deterministic benchmark matrix.
2. Score capability dimensions.
3. Run one real non-interactive CLI prompt if available.


```python
import shutil
for tool in ["openclaw", "codex", "claude", "opencode"]:
    print(tool, "available:", shutil.which(tool) is not None)
```

## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: executes a real provider/CLI flow with explicit graceful-skip behavior.


```python
# Deterministic Demo
capabilities = {
    "openclaw": {"repo_mutation": 1, "local_agent": 1, "policy_controls": 1},
    "codex": {"repo_mutation": 1, "local_agent": 0, "policy_controls": 1},
    "claude": {"repo_mutation": 1, "local_agent": 0, "policy_controls": 1},
    "opencode": {"repo_mutation": 1, "local_agent": 0, "policy_controls": 0},
}

for tool, dims in capabilities.items():
    score = sum(dims.values())
    print(tool, score, dims)

assert "openclaw" in capabilities
```

```python
# Live Demo
import shutil
import subprocess

candidates = [
    ["codex", "exec", "Explain in one sentence why code review is still required with coding agents."],
    ["claude", "-p", "Explain in one sentence why code review is still required with coding agents."],
    ["opencode", "run", "Explain in one sentence why code review is still required with coding agents."],
    ["openclaw", "agent", "--local", "--to", "+15555550123", "--message", "In one short sentence, why benchmark coding agents?", "--timeout", "60"],
]

for cmd in candidates:
    if shutil.which(cmd[0]) is None:
        continue
    try:
        p = subprocess.run(cmd, capture_output=True, text=True, timeout=70)
    except Exception as exc:
        print(f"command failed to execute ({cmd[0]}): {exc}")
        continue
    if p.returncode == 0:
        out = (p.stdout or p.stderr).strip()
        print("used:", " ".join(cmd[:2]))
        print(out[:1000])
        break
else:
    print("Skipping live CLI benchmark: no compatible non-interactive CLI found in runtime.")
```
