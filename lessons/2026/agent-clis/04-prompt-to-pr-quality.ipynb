## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: executes a real provider/CLI flow with explicit graceful-skip behavior.


```python
# Deterministic Demo
rubric = ["problem", "solution", "risks", "tests"]
summary = {
    "problem": "Notebook content was duplicated and low-signal.",
    "solution": "Rewrote lessons with unique progressive demos.",
    "risks": "Live paths depend on env/tool availability.",
    "tests": "Notebook structure and cleanliness checks pass.",
}

score = sum(1 for k in rubric if summary.get(k)) / len(rubric)
print("rubric score:", score)
assert score == 1.0
```

```python
# Live Demo
import shutil
import subprocess

prompt = "Write a PR summary with headings: Problem, Solution, Risks, Tests. Keep it concise."
cmds = [["codex", "exec", prompt], ["claude", "-p", prompt], ["opencode", "run", prompt]]
for cmd in cmds:
    if shutil.which(cmd[0]) is None:
        continue
    p = subprocess.run(cmd, capture_output=True, text=True, timeout=60)
    if p.returncode == 0:
        out = (p.stdout or p.stderr).strip()
        print(out[:1200])
        lower = out.lower()
        ok = all(h in lower for h in ["problem", "solution", "risks", "tests"])
        print("rubric headers present:", ok)
        break
else:
    print("Skipping live PR summary generation: no compatible non-interactive CLI available.")
```
