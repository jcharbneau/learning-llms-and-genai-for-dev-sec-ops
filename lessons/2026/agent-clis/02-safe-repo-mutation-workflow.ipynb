```python
# Deterministic Demo
workflow = [
    ("read_contracts", "auto"),
    ("inspect_git_status", "auto"),
    ("prepare_patch", "approval"),
    ("run_tests", "auto"),
    ("commit_changes", "approval"),
]

for step, gate in workflow:
    print(step, "->", gate)

approval_steps = [s for s, g in workflow if g == "approval"]
assert approval_steps == ["prepare_patch", "commit_changes"]
```

```python
# Live Demo
import shutil
import subprocess

prompt = "Draft a 6-step safe repository mutation workflow with explicit approval gates before write operations."
cmds = [
    ["openclaw", "agent", "--local", "--to", "+15555550123", "--message", prompt, "--timeout", "70"],
    ["codex", "exec", prompt],
    ["claude", "-p", prompt],
    ["opencode", "run", prompt],
]

for cmd in cmds:
    if shutil.which(cmd[0]) is None:
        continue
    p = subprocess.run(cmd, capture_output=True, text=True, timeout=60)
    if p.returncode == 0:
        out = (p.stdout or p.stderr).strip()
        print(out[:1200])
        assert "approval" in out.lower() or "review" in out.lower()
        break
else:
    print("Skipping live workflow draft: no compatible non-interactive CLI available.")
```
