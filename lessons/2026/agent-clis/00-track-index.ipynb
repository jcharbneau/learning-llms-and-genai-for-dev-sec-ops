# 2026 Agent CLIs Lessons

## Track Purpose
Operational playbooks for OpenCode, Codex, and Claude Code

## Lesson Roadmap
- `01-opencode-codex-claudecode-playbook.ipynb`: CLI capability playbook - Compare core capabilities and safety defaults.
- `02-safe-repo-mutation-workflow.ipynb`: Safe repo mutation workflow - Apply branch/test/commit gates for agent changes.
- `03-approval-and-sandbox-policies.ipynb`: Approval and sandbox policies - Control risk for shell and file operations.
- `04-prompt-to-pr-quality.ipynb`: Prompt-to-PR quality patterns - Evaluate output quality and maintainability.
- `05-resilience-under-failure.ipynb`: Resilience under failure - Handle flaky tools, network faults, and partial runs.
- `06-security-posture-checks.ipynb`: Security posture checks - Prevent secret leaks and unsafe commands.
- `07-latency-and-cost-benchmarking.ipynb`: Latency and cost benchmarking - Compare throughput/cost across runtimes.
- `08-team-adoption-playbook.ipynb`: Team adoption playbook - Standardize workflows, review policies, and auditability.


## How To Use This Track
1. Start with `01` for baseline.
2. Move through reliability/ops/security themes.
3. Capture tradeoffs and adoption decisions per lesson.
