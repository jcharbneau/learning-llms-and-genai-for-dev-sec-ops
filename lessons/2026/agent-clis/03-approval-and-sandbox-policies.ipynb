```python
# Deterministic Demo
commands = [
    "git status",
    "docker compose up -d",
    "git push origin main",
    "rm -rf /",
]

def classify(cmd: str) -> str:
    if "rm -rf /" in cmd:
        return "deny"
    if cmd.startswith("git push") or cmd.startswith("docker compose up"):
        return "approve"
    return "auto"

labels = {c: classify(c) for c in commands}
print(labels)
assert labels["git status"] == "auto"
```

```python
# Live Demo
import shutil
import subprocess

prompt = "Propose a concise auto/approve/deny command policy for AI coding agents in markdown table format."
cmds = [
    ["openclaw", "agent", "--local", "--to", "+15555550123", "--message", prompt, "--timeout", "70"],
    ["codex", "exec", prompt],
    ["claude", "-p", prompt],
    ["opencode", "run", prompt],
]
for cmd in cmds:
    if shutil.which(cmd[0]) is None:
        continue
    p = subprocess.run(cmd, capture_output=True, text=True, timeout=60)
    if p.returncode == 0:
        out = (p.stdout or p.stderr).strip()
        print(out[:1200])
        assert "approve" in out.lower() or "deny" in out.lower()
        break
else:
    print("Skipping live policy generation: no compatible non-interactive CLI available.")
```
