```python
# Deterministic Demo
from string import Template
t=Template('Service: $service\nIssue: $issue')
out=t.substitute(service='billing-api', issue='timeout spike')
print(out)
assert 'billing-api' in out
```

```python
# Live Demo
import os
try:
    from langchain_core.prompts import ChatPromptTemplate
    from langchain_openai import ChatOpenAI
except Exception as exc:
    print(f"Skipping live run: dependency unavailable ({exc})")
else:
    if not os.getenv("OPENAI_API_KEY"):
        print("Skipping live run: OPENAI_API_KEY not set.")
    else:
        prompt = ChatPromptTemplate.from_template('Explain {topic} for {audience} in one sentence.')
        llm = ChatOpenAI(model='gpt-4.1-mini', temperature=0, timeout=20)
        chain = prompt | llm
        out = chain.invoke({'topic': 'prompt templating', 'audience': 'SRE engineers'})
        print(out.content if hasattr(out, 'content') else out)
```
