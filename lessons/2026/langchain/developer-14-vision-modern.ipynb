```python
# Live Demo
import os
try:
    from langchain_openai import AzureChatOpenAI
except Exception as exc:
    print(f"Skipping live vision run: dependency unavailable ({exc})")
else:
    req = ["AZURE_OPENAI_API_KEY", "AZURE_OPENAI_ENDPOINT", "AZURE_VISION_DEPLOYMENT"]
    missing = [k for k in req if not os.getenv(k)]
    if missing:
        print(f"Skipping live vision run: missing env vars {missing}")
    else:
        llm = AzureChatOpenAI(
            azure_endpoint=os.getenv("AZURE_OPENAI_ENDPOINT"),
            api_key=os.getenv("AZURE_OPENAI_API_KEY"),
            azure_deployment=os.getenv("AZURE_VISION_DEPLOYMENT"),
            api_version="2024-02-15-preview",
            temperature=0,
        )
        out = llm.invoke('One sentence: what makes multimodal payloads safe?')
        print(out.content if hasattr(out, 'content') else out)
```
