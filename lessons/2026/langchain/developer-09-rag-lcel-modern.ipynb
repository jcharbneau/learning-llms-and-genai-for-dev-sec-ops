```python
# Live Demo
import os
try:
    from langchain_core.prompts import ChatPromptTemplate
    from langchain_core.runnables import RunnablePassthrough
    from langchain_openai import ChatOpenAI
except Exception as exc:
    print(f"Skipping live LCEL run: dependency unavailable ({exc})")
else:
    if not os.getenv("OPENAI_API_KEY"):
        print("Skipping live LCEL run: OPENAI_API_KEY not set.")
    else:
        prompt = ChatPromptTemplate.from_template('Context: {context}\nQuestion: {question}\nAnswer briefly.')
        llm = ChatOpenAI(model='gpt-4.1-mini', temperature=0, timeout=20)
        chain = ({'context': RunnablePassthrough(), 'question': RunnablePassthrough()} | prompt | llm)
        out = chain.invoke('RAG grounding context')
        print(out.content if hasattr(out, 'content') else out)
```
