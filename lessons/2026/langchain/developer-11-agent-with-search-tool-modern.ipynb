```python
# Live Demo
import os
try:
    from langchain.tools import DuckDuckGoSearchRun
    from langchain_openai import ChatOpenAI
except Exception as exc:
    print(f"Skipping live agent run: dependency unavailable ({exc})")
else:
    if not os.getenv("OPENAI_API_KEY"):
        print("Skipping live agent run: OPENAI_API_KEY not set.")
    else:
        try:
            search = DuckDuckGoSearchRun()
            snippet = search.invoke('best practice for prompt injection defense')
            llm = ChatOpenAI(model='gpt-4.1-mini', temperature=0, timeout=20)
            out = llm.invoke(f'Summarize this snippet in one sentence: {snippet[:500]}')
            print(out.content if hasattr(out, 'content') else out)
        except Exception as exc:
            print(f"Skipping live search flow: {exc}")
```
