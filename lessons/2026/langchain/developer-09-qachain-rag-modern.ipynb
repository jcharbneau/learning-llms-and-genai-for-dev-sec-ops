```python
# Live Demo
import os
try:
    from langchain_core.documents import Document
    from langchain_community.vectorstores import FAISS
    from langchain_openai import OpenAIEmbeddings, ChatOpenAI
except Exception as exc:
    print(f"Skipping live RAG run: dependency unavailable ({exc})")
else:
    if not os.getenv("OPENAI_API_KEY"):
        print("Skipping live RAG run: OPENAI_API_KEY not set.")
    else:
        docs = [Document(page_content="RAG grounds answers using retrieved context.")]
        vs = FAISS.from_documents(docs, OpenAIEmbeddings())
        context = vs.as_retriever(search_kwargs={"k":1}).invoke('How does RAG improve trust?')[0].page_content
        llm = ChatOpenAI(model='gpt-4.1-mini', temperature=0, timeout=20)
        out = llm.invoke(f"Context: {context}\nQuestion: How does RAG improve trust?\nAnswer in one sentence.")
        print(out.content if hasattr(out, 'content') else out)
```
