```python
# Deterministic Demo
import hashlib
cache={}
prompt="Summarize runbook update in one sentence."
key=hashlib.sha256(prompt.encode()).hexdigest()
cache.setdefault(key,"value-v1")
cache.setdefault(key,"value-v2")
print(cache[key])
assert cache[key]=="value-v1"
```

```python
# Live Demo
import os, hashlib
try:
    from langchain_openai import ChatOpenAI
except Exception as exc:
    print(f"Skipping live cache run: dependency unavailable ({exc})")
else:
    if not os.getenv("OPENAI_API_KEY"):
        print("Skipping live cache run: OPENAI_API_KEY not set.")
    else:
        llm = ChatOpenAI(model='gpt-4.1-mini', temperature=0, timeout=20)
        memo = {}
        prompt = 'One sentence: why cache invalidation is hard.'
        key = hashlib.sha256(prompt.encode()).hexdigest()
        if key not in memo:
            memo[key] = llm.invoke(prompt).content
            print('cache miss')
        else:
            print('cache hit')
        print(memo[key])
```
