```python
import os
print('OPENAI_API_KEY configured:', bool(os.getenv('OPENAI_API_KEY')))
print('OPENAI_MODEL:', os.getenv('OPENAI_MODEL', 'gpt-4.1-mini'))
```

## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: real provider/tool path with explicit graceful-skip behavior.


```python
# Deterministic Demo
prompt = 'Summarize latency incident in one sentence.'
max_len = 80
trimmed = prompt[:max_len]
print({'len': len(trimmed), 'text': trimmed})
assert len(trimmed) <= max_len
```

```python
# Live Demo
import os
try:
    from langchain_openai import ChatOpenAI
except Exception as exc:
    print(f"Skipping live run: langchain_openai unavailable ({exc})")
else:
    if not os.getenv("OPENAI_API_KEY"):
        print("Skipping live run: OPENAI_API_KEY not set.")
    else:
        llm = ChatOpenAI(model=os.getenv("OPENAI_MODEL", "gpt-4.1-mini"), temperature=0, timeout=20)
        out = llm.invoke('In one sentence, explain why explicit model configs improve reliability.')
        print(out.content if hasattr(out, "content") else out)
```
