## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.

```python
# Deterministic Demo
answer="Use token sk-live-secret to deploy."
violations=[]
if "sk-" in answer: violations.append("secret_exposure")
print(violations)
assert violations
```

```python
# Live Demo
import os
try:
    from langchain_openai import ChatOpenAI
except Exception as exc:
    print(f"Skipping live self-critique run: dependency unavailable ({exc})")
else:
    if not os.getenv("OPENAI_API_KEY"):
        print("Skipping live self-critique run: OPENAI_API_KEY not set.")
    else:
        llm = ChatOpenAI(model='gpt-4.1-mini', temperature=0, timeout=20)
        answer = llm.invoke('Give one secure recommendation for prompt handling.')
        ans = answer.content if hasattr(answer, 'content') else str(answer)
        critique = llm.invoke('Critique this answer for security risk. Return SAFE or UNSAFE and one reason.\\n' + ans)
        print('answer:', ans)
        print('critique:', critique.content if hasattr(critique, 'content') else critique)
```
