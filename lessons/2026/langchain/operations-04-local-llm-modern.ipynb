## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: real provider/tool path with explicit graceful-skip behavior.


```python
# Deterministic Demo
from urllib.parse import urlparse
import os
base=os.getenv("OLLAMA_BASE_URL", "http://host.docker.internal:11434")
u=urlparse(base)
print({"host":u.hostname,"port":u.port})
assert u.hostname in {"host.docker.internal","localhost","127.0.0.1"}
```

```python
# Live Demo
import os
try:
    from langchain_community.chat_models import ChatOllama
except Exception as exc:
    print(f"Skipping live local run: dependency unavailable ({exc})")
else:
    base = os.getenv("OLLAMA_BASE_URL", "http://host.docker.internal:11434")
    model = os.getenv("OLLAMA_MODEL", "qwen2.5-coder:1.5b")
    try:
        llm = ChatOllama(base_url=base, model=model, temperature=0)
        out = llm.invoke('One sentence: when should teams prefer local LLMs?')
        print(out.content if hasattr(out, 'content') else out)
    except Exception as exc:
        print(f"Skipping live local run: {exc}")
```
