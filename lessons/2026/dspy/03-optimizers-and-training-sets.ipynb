```python
# Deterministic Demo
eval_set = [
    {"q": "What is RAG?", "expected_keyword": "retrieval"},
    {"q": "Why cache model calls?", "expected_keyword": "cost"},
]

candidate_scores = {
    "program_a": [0.90, 0.45],
    "program_b": [0.82, 0.79],
    "program_c": [0.68, 0.88],
}

avg = {k: sum(v) / len(v) for k, v in candidate_scores.items()}
best = max(avg, key=avg.get)
print("avg:", avg, "best:", best)
assert best in candidate_scores
```

```python
# Live Demo
import os

try:
    import dspy
except Exception as exc:
    print(f"Skipping live optimization preview: dspy unavailable ({exc})")
else:
    api_key = os.getenv("OPENAI_API_KEY")
    if not api_key:
        print("Skipping live optimization preview: OPENAI_API_KEY not set.")
    else:
        dspy.configure(lm=dspy.LM("openai/gpt-4.1-mini", api_key=api_key, temperature=0))
        predict = dspy.Predict("question -> answer")
        out = predict(question="Why does retrieval improve factual reliability?")
        print(out.answer)
        assert out.answer
```
