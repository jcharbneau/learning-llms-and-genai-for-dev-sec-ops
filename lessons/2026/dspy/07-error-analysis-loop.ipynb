## How It Works

2. Prioritize by frequency.
3. Generate live remediation suggestions for observed failures.

## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: executes a real provider/CLI flow with explicit graceful-skip behavior.


```python
# Deterministic Demo
failures = [
    "missing_field:owner",
    "hallucination:unsupported_fact",
    "missing_field:priority",
    "format_error:invalid_json",
]

remediation = {
    "missing_field": "tighten schema and add validator",
    "hallucination": "increase grounding and retrieval constraints",
    "format_error": "use structured parser with retry",
}

counts = {}
for f in failures:
    kind = f.split(":", 1)[0]
    counts[kind] = counts.get(kind, 0) + 1

ordered = sorted(counts.items(), key=lambda kv: kv[1], reverse=True)
print(ordered)
for kind, n in ordered:
    print(kind, n, "->", remediation[kind])
assert ordered[0][0] == "missing_field"
```

```python
# Live Demo
import os

try:
    import dspy
except Exception as exc:
    print(f"Skipping live remediation demo: dspy unavailable ({exc})")
else:
    api_key = os.getenv("OPENAI_API_KEY")
    if not api_key:
        print("Skipping live remediation demo: OPENAI_API_KEY not set.")
    else:
        dspy.configure(lm=dspy.LM("openai/gpt-4.1-mini", api_key=api_key, temperature=0))
        suggest = dspy.Predict("failure_text -> remediation")
        for f in ["missing required field owner", "output not valid json"]:
            out = suggest(failure_text=f)
            print(f"{f} -> {out.remediation}")
            assert out.remediation
```
