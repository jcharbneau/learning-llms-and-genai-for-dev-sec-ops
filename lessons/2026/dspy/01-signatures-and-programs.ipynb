# 01 Signatures and Typed Contracts (DSPy, 2026)

## What This Lesson Is
Use typed signatures to define stable program interfaces for LLM behaviors.

## Scientific Lens
- Concept: Type-constrained prompt programming interfaces
- Measure: Signature conformance rate and output completeness
- Validity Limit: Type-conformant outputs may still be semantically incorrect.


## How It Works
1. Define typed contracts deterministically.
2. Validate signature shape before runtime.
3. Run live `dspy.Predict` with configured LM.

## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: executes a real provider/CLI flow with explicit graceful-skip behavior.


```python
# Deterministic Demo
from dataclasses import dataclass

@dataclass
class ExplainRiskOutput:
    topic: str
    explanation: str

sample = ExplainRiskOutput(topic="prompt injection", explanation="Untrusted prompt content can manipulate model behavior.")
print(sample)
assert sample.topic and sample.explanation
```

```python
# Live Demo
import os

try:
    import dspy
except Exception as exc:
    print(f"Skipping live DSPy run: dspy unavailable ({exc})")
else:
    api_key = os.getenv("OPENAI_API_KEY")
    if not api_key:
        print("Skipping live DSPy run: OPENAI_API_KEY not set.")
    else:
        lm = dspy.LM("openai/gpt-4.1-mini", api_key=api_key, temperature=0)
        dspy.configure(lm=lm)

        class ExplainRisk(dspy.Signature):
            topic: str = dspy.InputField()
            explanation: str = dspy.OutputField()

        predict = dspy.Predict(ExplainRisk)
        out = predict(topic="prompt injection")
        print(out.explanation)
        assert out.explanation
```
