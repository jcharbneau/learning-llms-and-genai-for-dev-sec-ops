# DSPy: Signatures and Programs (2026)

Goal: introduce DSPy-style typed signatures and programmable prompt workflows.


```python
try:
    import dspy
    HAS_DSPY = True
except Exception as exc:
    HAS_DSPY = False
    DSPY_IMPORT_ERROR = str(exc)

print(f"dspy available: {HAS_DSPY}")
```

## Minimal signature example
This uses a tiny module and checks that structured fields are returned.


```python
if not HAS_DSPY:
    print("Skipping: dspy is not installed. Install with: pip install dspy-ai")
    print(f"Import error: {DSPY_IMPORT_ERROR}")
else:
    class Summarize(dspy.Signature):
        """Summarize input text in one sentence."""
        text: str = dspy.InputField()
        summary: str = dspy.OutputField()

    # Note: model backend config is environment-specific and omitted by default here.
    print("DSPy signature declared successfully.")
    assert hasattr(Summarize, "__name__")
```
