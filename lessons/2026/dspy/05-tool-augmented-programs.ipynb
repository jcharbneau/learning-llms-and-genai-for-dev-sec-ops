```python
# Deterministic Demo
def safe_calc(expr: str) -> str:
    return str(eval(expr, {"__builtins__": {}}, {}))

def local_lookup(topic: str) -> str:
    kb = {"rag": "Retrieval-augmented generation grounds answers with retrieved context."}
    return kb.get(topic.lower(), "unknown topic")

print("calc:", safe_calc("12 * 7"))
print("lookup:", local_lookup("rag"))
assert safe_calc("12*7") == "84"
```

```python
# Live Demo
import os

try:
    import dspy
except Exception as exc:
    print(f"Skipping live tool-augmented demo: dspy unavailable ({exc})")
else:
    api_key = os.getenv("OPENAI_API_KEY")
    if not api_key:
        print("Skipping live tool-augmented demo: OPENAI_API_KEY not set.")
    else:
        dspy.configure(lm=dspy.LM("openai/gpt-4.1-mini", api_key=api_key, temperature=0))
        planner = dspy.Predict("question -> tool_plan")
        plan = planner(question="Compute 13*9 and explain why deterministic tools are safer for arithmetic.")
        print("plan:", plan.tool_plan)
        print("tool_result:", 13 * 9)
        assert (13 * 9) == 117
```
