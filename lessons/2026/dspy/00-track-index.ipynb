# 2026 DSPy Lessons

## Track Purpose
Programmatic prompting and optimizer-driven workflows

## Lesson Roadmap
- `01-signatures-and-programs.ipynb`: Signatures and programs - Define typed interfaces for language programs.
- `02-composable-modules.ipynb`: Composable modules - Compose reusable DSPy modules for complex flows.
- `03-optimizers-and-training-sets.ipynb`: Optimizers and training sets - Improve behavior using small curated examples.
- `04-constraints-and-structured-output.ipynb`: Constraints and structured output - Enforce output contracts reliably.
- `05-tool-augmented-programs.ipynb`: Tool-augmented programs - Combine tools and signatures in one program.
- `06-regression-testing.ipynb`: Regression testing - Protect quality with deterministic checks.
- `07-error-analysis-loop.ipynb`: Error analysis loop - Use failures to iteratively improve programs.
- `08-production-hand-off.ipynb`: Production hand-off - Version and operationalize DSPy programs.


## How To Use This Track
1. Start with `01` for baseline.
2. Move through reliability/ops/security themes.
3. Capture tradeoffs and adoption decisions per lesson.
