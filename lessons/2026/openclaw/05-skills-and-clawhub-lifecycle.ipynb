## How It Works
1. Model skill precedence resolution rules and detect shadowing risks.
2. Validate skill metadata and expected command/tool dependencies deterministically.
3. Run live skill discovery/install/update commands where available.


```python
import os
import shutil

HAS_OPENCLAW = shutil.which("openclaw") is not None
print("openclaw available:", HAS_OPENCLAW)
print("OPENAI_API_KEY configured:", bool(os.getenv("OPENAI_API_KEY")))
print("OLLAMA_BASE_URL:", os.getenv("OLLAMA_BASE_URL", "<unset>"))
```

## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: uses real OpenClaw integration commands and skips gracefully if prerequisites are missing.


```python
# Deterministic Demo
candidates = [
    {"name":"github-review","location":"bundled"},
    {"name":"github-review","location":"workspace"},
    {"name":"release-notes","location":"local"},
]
precedence = {"workspace":3, "local":2, "bundled":1}
resolved = {}
for s in candidates:
    cur = resolved.get(s["name"])
    if cur is None or precedence[s["location"]] > precedence[cur["location"]]:
        resolved[s["name"]] = s
print(resolved)
assert resolved["github-review"]["location"] == "workspace"
```

```python
# Live Demo
import shutil, subprocess

if not HAS_OPENCLAW:
    print("Skipping live skills demo: openclaw CLI not installed.")
else:
    cmds = [["openclaw", "skills", "list"], ["openclaw", "skills", "reload"]]
    has_clawhub = shutil.which("clawhub") is not None
    if has_clawhub:
        cmds.append(["clawhub", "search", "review"])
    else:
        print("clawhub not installed; skipping ClawHub search command.")
    for cmd in cmds:
        print("$", " ".join(cmd))
        p = subprocess.run(cmd, capture_output=True, text=True)
        print((p.stdout or p.stderr).strip()[:1400])
```
