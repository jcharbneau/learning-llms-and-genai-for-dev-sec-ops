## Scientific Lens
- Concept: Gateway state preflight and command contract validation.
- Measure: Readiness score = checks passed / checks total.
- Validity Limit: Preflight confirms availability, not provider response quality.


## How It Works
1. Run deterministic checks for required CLI commands and parseable status fields.
2. Compute readiness score for onboarding gates.
3. Run live `openclaw` commands when available and show actionable remediation when missing.


```python
import os
import shutil

HAS_OPENCLAW = shutil.which("openclaw") is not None
print("openclaw available:", HAS_OPENCLAW)
print("OPENAI_API_KEY configured:", bool(os.getenv("OPENAI_API_KEY")))
print("OLLAMA_BASE_URL:", os.getenv("OLLAMA_BASE_URL", "<unset>"))
```

## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: runs real OpenClaw CLI operations when available; otherwise prints explicit skip guidance.


```python
# Deterministic Demo
required_commands = ["openclaw --help", "openclaw gateway status", "openclaw models list"]
simulated_help = "Usage: openclaw [command]"
simulated_status = "Service: systemd (disabled) | Gateway: bind=loopback (127.0.0.1), port=18789 | Dashboard: http://127.0.0.1:18789/"

checks = {
    "help_text_present": "Usage: openclaw" in simulated_help,
    "gateway_bind_present": "Gateway:" in simulated_status,
    "dashboard_present": "Dashboard:" in simulated_status,
    "command_count_ok": len(required_commands) == 3,
}
readiness = sum(int(v) for v in checks.values()) / len(checks)
print("checks:", checks)
print("readiness:", readiness)
assert readiness >= 0.75
```

```python
# Live Demo
import shutil
import subprocess

if shutil.which("openclaw") is None:
    print("Skipping live gateway demo: openclaw CLI is not installed.")
else:
    for cmd in (["openclaw", "--help"], ["openclaw", "gateway", "status"], ["openclaw", "models", "list"]):
        print("$", " ".join(cmd))
        proc = subprocess.run(cmd, capture_output=True, text=True)
        out = (proc.stdout or proc.stderr).strip()
        print(out[:1200] if out else "<no output>")
```
