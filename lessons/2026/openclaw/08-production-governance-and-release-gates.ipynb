# 08 Production Governance and Release Gates (OpenClaw, 2026)

## What This Lesson Is
Create production handoff criteria for OpenClaw deployments with measurable readiness and rollback signals.

## Scientific Lens
- Concept: Release gating with explicit operational, security, and quality controls.
- Measure: Gate pass score across governance checklist dimensions.
- Validity Limit: Checklist pass does not eliminate unknown unknowns in dynamic production environments.


## How It Works
1. Define deterministic release gates across reliability, safety, and observability.
2. Compute weighted readiness score and enforce pass threshold.
3. Collect runtime snapshots from OpenClaw to attach to release artifacts.


```python
import os
import shutil

HAS_OPENCLAW = shutil.which("openclaw") is not None
print("openclaw available:", HAS_OPENCLAW)
print("OPENAI_API_KEY configured:", bool(os.getenv("OPENAI_API_KEY")))
print("OLLAMA_BASE_URL:", os.getenv("OLLAMA_BASE_URL", "<unset>"))
```

## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: runs real OpenClaw CLI operations when available; otherwise prints explicit skip guidance.


```python
# Deterministic Demo
gates = {
    "slo_defined": 1.0,
    "rollback_runbook": 1.0,
    "alerting_wired": 0.5,
    "policy_review_signed": 1.0,
    "cost_guardrail_set": 0.5,
}
score = sum(gates.values()) / len(gates)
print("release_score:", score)
assert score >= 0.7
```

```python
# Live Demo
import json
import shutil
import subprocess

snapshot = {}
if shutil.which("openclaw") is None:
    print("Skipping live governance demo: openclaw CLI is not installed.")
else:
    for key, cmd in {
        "gateway_status": ["openclaw", "gateway", "status"],
        "models_list": ["openclaw", "models", "list"],
    }.items():
        proc = subprocess.run(cmd, capture_output=True, text=True)
        snapshot[key] = (proc.stdout or proc.stderr).strip()[:1200]
    print(json.dumps(snapshot, indent=2))
```
