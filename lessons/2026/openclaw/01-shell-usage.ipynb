## Gateway Token Flow (Get/Set)
The Python API-based lessons (`02..09`) call OpenClaw's `/v1/chat/completions` endpoint.
That endpoint expects a gateway token.

This flow gives you a repeatable way to get/set it in notebook runtime:
1. Read existing token from env.
2. If missing, generate one for local lab usage.
3. Set it in current kernel env so subsequent cells can run.


```python
import os
import secrets

current = os.getenv("OPENCLAW_GATEWAY_TOKEN", "")
print("Token already configured:", bool(current))

if not current:
    current = f"ocw-{secrets.token_urlsafe(24)}"
    os.environ["OPENCLAW_GATEWAY_TOKEN"] = current
    print("Generated local token for this notebook session.")

print("OPENCLAW_GATEWAY_TOKEN length:", len(current))
print("Persist for Docker runs by adding to .env:")
print("OPENCLAW_GATEWAY_TOKEN=<your-token>")
```
