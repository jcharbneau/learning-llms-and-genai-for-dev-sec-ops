# 2026 OpenClaw Track

This track teaches OpenClaw as an AI operations platform: gateway setup, provider routing, channels, skills, automation, and production governance.

## Sequence
- `01-shell-usage.ipynb`
- `02-foundations-gateway-models.ipynb`
- `03-provider-routing-and-failover.ipynb`
- `04-channel-routing-and-session-boundaries.ipynb`
- `05-web-tools-and-browser-automation.ipynb`
- `06-skills-and-clawhub-lifecycle.ipynb`
- `07-cron-and-webhook-automations.ipynb`
- `08-showcase-integration-patterns.ipynb`
- `09-production-hardening-and-release-gates.ipynb`

## Why this track
OpenClaw is not just a chat shell. It is a gateway for multi-channel agent execution, policy, and automation. These lessons focus on AI system behavior and integration methodology, not generic Python tutorials.
