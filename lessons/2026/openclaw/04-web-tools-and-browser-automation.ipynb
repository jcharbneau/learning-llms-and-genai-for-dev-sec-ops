## What This Lesson Is
Combine `web_search`/`web_fetch` with browser automation to handle dynamic web workflows safely.

## Scientific Lens
- Concept: Choose the minimal web capability needed: search/fetch first, browser automation when necessary.
- Measure: Task completion rate by tool choice under static vs JS-heavy targets.
- Validity Limit: Browser automation introduces higher complexity and security risk than fetch/search.


## How It Works
1. Classify tasks into fetch/search vs browser categories based on site behavior.
2. Model security controls for browser tool usage and evaluate risk exposure.
3. Configure web tool settings and inspect browser status through OpenClaw CLI.


```python
import os
import shutil

HAS_OPENCLAW = shutil.which("openclaw") is not None
print("openclaw available:", HAS_OPENCLAW)
print("OPENAI_API_KEY configured:", bool(os.getenv("OPENAI_API_KEY")))
print("OLLAMA_BASE_URL:", os.getenv("OLLAMA_BASE_URL", "<unset>"))
```

## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: uses real OpenClaw integration commands and skips gracefully if prerequisites are missing.


```python
# Deterministic Demo
tasks = [
    {"name":"read docs page","js_heavy":False,"requires_login":False},
    {"name":"post status update","js_heavy":True,"requires_login":True},
    {"name":"grab API pricing","js_heavy":False,"requires_login":False},
]

def select_tool(t):
    if t["requires_login"] or t["js_heavy"]:
        return "browser"
    return "web_fetch_or_search"

choices = {t["name"]: select_tool(t) for t in tasks}
print(choices)
assert choices["post status update"] == "browser"
assert choices["read docs page"] == "web_fetch_or_search"
```

```python
# Live Demo
import shutil, subprocess

if not HAS_OPENCLAW:
    print("Skipping live web-tools demo: openclaw CLI not installed.")
else:
    cmds = [
        ["openclaw", "configure", "--section", "web"],
        ["openclaw", "browser", "status"],
        ["openclaw", "browser", "tabs"],
    ]
    for cmd in cmds:
        print("$", " ".join(cmd))
        p = subprocess.run(cmd, capture_output=True, text=True)
        print((p.stdout or p.stderr).strip()[:1400])
```
