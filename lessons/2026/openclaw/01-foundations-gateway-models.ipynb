## What This Lesson Is
Establish a reliable OpenClaw baseline: gateway health, dashboard access, and explicit model selection.

## Scientific Lens
- Concept: Deterministic readiness gates reduce downstream setup failures.
- Measure: Readiness ratio = passed checks / total critical checks.
- Validity Limit: Readiness does not guarantee model quality or external provider uptime.


## How It Works
1. Validate gateway, dashboard, and models command surface before any workflow build.
2. Enforce provider/model naming contract (`provider/model`) using deterministic checks.
3. Run live status commands and capture actionable outputs for operator handoff.


```python
import os
import shutil

HAS_OPENCLAW = shutil.which("openclaw") is not None
print("openclaw available:", HAS_OPENCLAW)
print("OPENAI_API_KEY configured:", bool(os.getenv("OPENAI_API_KEY")))
print("OLLAMA_BASE_URL:", os.getenv("OLLAMA_BASE_URL", "<unset>"))
```

## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: uses real OpenClaw integration commands and skips gracefully if prerequisites are missing.


```python
# Deterministic Demo
required = {
    "provider_model_format": "openai/gpt-5.1-codex".count("/") == 1,
    "dashboard_url_shape": "http://127.0.0.1:18789/".startswith("http://"),
    "command_surface": all(k in {"status", "dashboard", "models list"} for k in ["status", "dashboard"]),
    "health_contract": True,
}
score = sum(int(v) for v in required.values()) / len(required)
print(required)
print("readiness_score=", score)
assert score >= 0.75
```

```python
# Live Demo
import shutil, subprocess

if not HAS_OPENCLAW:
    print("Skipping live foundations demo: openclaw CLI not installed.")
else:
    cmds = [
        ["openclaw", "status"],
        ["openclaw", "dashboard"],
        ["openclaw", "models", "status"],
        ["openclaw", "models", "list"],
    ]
    for cmd in cmds:
        print("$", " ".join(cmd))
        p = subprocess.run(cmd, capture_output=True, text=True)
        print((p.stdout or p.stderr).strip()[:1400])
```
