# 03 Channel Routing and Session Boundaries (OpenClaw, 2026)

## What This Lesson Is
Design multi-channel agent behavior with explicit routing and session-isolation boundaries.

## Scientific Lens
- Concept: Channel-aware routing prevents context bleed and improves operational safety.
- Measure: Isolation integrity across concurrent sender/channel sessions.
- Validity Limit: Channel metadata quality and pairing hygiene determine practical reliability.


## How It Works
1. Model routing decisions from channel/sender metadata to distinct agent sessions.
2. Validate no cross-session memory leakage in deterministic simulation.
3. Use live pairing/status commands to inspect channel readiness and policy state.


```python
import os
import shutil

HAS_OPENCLAW = shutil.which("openclaw") is not None
print("openclaw available:", HAS_OPENCLAW)
print("OPENAI_API_KEY configured:", bool(os.getenv("OPENAI_API_KEY")))
print("OLLAMA_BASE_URL:", os.getenv("OLLAMA_BASE_URL", "<unset>"))
```

## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: uses real OpenClaw integration commands and skips gracefully if prerequisites are missing.


```python
# Deterministic Demo
events = [
    {"channel":"telegram","sender":"alice","intent":"pr_review"},
    {"channel":"discord","sender":"team-bot","intent":"incident_triage"},
    {"channel":"slack","sender":"ops-oncall","intent":"daily_brief"},
]
route_map = {
    "pr_review": "agent:review",
    "incident_triage": "agent:security",
    "daily_brief": "agent:ops",
}

sessions = [f"{e['channel']}::{e['sender']}::{route_map[e['intent']]}" for e in events]
print(sessions)
assert len(set(sessions)) == len(events)
assert all("::agent:" in s for s in sessions)
```

```python
# Live Demo
import shutil, subprocess

if not HAS_OPENCLAW:
    print("Skipping live channel demo: openclaw CLI not installed.")
else:
    cmds = [
        ["openclaw", "pairing", "status"],
        ["openclaw", "status"],
    ]
    for cmd in cmds:
        print("$", " ".join(cmd))
        p = subprocess.run(cmd, capture_output=True, text=True)
        print((p.stdout or p.stderr).strip()[:1400])
```
