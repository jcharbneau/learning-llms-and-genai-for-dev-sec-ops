```python
# Deterministic Demo
candidates = [
    {"id": "A", "lexical": 0.92, "semantic": 0.40},
    {"id": "B", "lexical": 0.45, "semantic": 0.88},
    {"id": "C", "lexical": 0.63, "semantic": 0.61},
]

w_lex, w_sem = 0.45, 0.55
for c in candidates:
    c["hybrid"] = round((w_lex * c["lexical"]) + (w_sem * c["semantic"]), 3)

ranked = sorted(candidates, key=lambda x: x["hybrid"], reverse=True)
print(ranked)
assert ranked[0]["id"] in {"A", "B", "C"}
```

```python
# Live Demo
try:
    from llama_index.core import Document, VectorStoreIndex
except Exception as exc:
    print(f"Skipping live hybrid retrieval demo: dependency missing ({exc})")
else:
    docs = [
        Document(text="Lexical retrieval is strong for exact term matching."),
        Document(text="Semantic retrieval captures conceptual similarity."),
        Document(text="Hybrid retrieval combines lexical and semantic strengths."),
    ]
    idx = VectorStoreIndex.from_documents(docs)
    nodes = idx.as_retriever(similarity_top_k=2).retrieve("Why combine lexical and semantic retrieval?")
    for n in nodes:
        print("score:", getattr(n, "score", None), "text:", n.text[:110])
    assert len(nodes) >= 1
```
