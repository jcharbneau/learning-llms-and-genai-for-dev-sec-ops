```python
# Deterministic Demo
results = [
    {"doc": "D1", "first_pass": 0.89, "rerank": 0.42},
    {"doc": "D2", "first_pass": 0.84, "rerank": 0.93},
    {"doc": "D3", "first_pass": 0.81, "rerank": 0.56},
]

before = [r["doc"] for r in sorted(results, key=lambda x: x["first_pass"], reverse=True)]
after = [r["doc"] for r in sorted(results, key=lambda x: x["rerank"], reverse=True)]

print("before:", before)
print("after:", after)
assert before[0] != after[0]
```

```python
# Live Demo
try:
    from llama_index.core import Document, VectorStoreIndex
except Exception as exc:
    print(f"Skipping live reranking demo: dependency missing ({exc})")
else:
    docs = [
        Document(text="First-pass retrieval optimizes recall."),
        Document(text="Reranking prioritizes precision for top results."),
        Document(text="Quality evaluation should compare before and after reranking."),
    ]
    idx = VectorStoreIndex.from_documents(docs)
    ans = str(idx.as_query_engine(similarity_top_k=3).query("Why does reranking help?"))
    print(ans)
    assert ans
```
