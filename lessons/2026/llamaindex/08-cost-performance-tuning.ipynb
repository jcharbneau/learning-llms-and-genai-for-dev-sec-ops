```python
# Deterministic Demo
configs = [
    {"name": "k2_chunk400", "quality": 0.71, "latency_ms": 520, "cost": 0.010},
    {"name": "k4_chunk300", "quality": 0.82, "latency_ms": 780, "cost": 0.018},
    {"name": "k6_chunk250", "quality": 0.85, "latency_ms": 1120, "cost": 0.026},
]

for c in configs:
    c["utility"] = round(c["quality"] - (c["latency_ms"] / 5000) - (c["cost"] * 8), 3)

best = sorted(configs, key=lambda x: x["utility"], reverse=True)[0]
print(configs)
print("best:", best)
assert "name" in best
```

```python
# Live Demo
try:
    from llama_index.core import Document, VectorStoreIndex
except Exception as exc:
    print(f"Skipping live tuning demo: dependency missing ({exc})")
else:
    docs = [
        Document(text="Higher top_k may improve recall but adds latency."),
        Document(text="Smaller chunks may improve focus while increasing total nodes."),
        Document(text="Reranking improves precision with extra compute cost."),
    ]
    idx = VectorStoreIndex.from_documents(docs)
    query = "What tradeoff does top_k introduce?"
    for top_k in [1, 2, 3]:
        ans = str(idx.as_query_engine(similarity_top_k=top_k).query(query))
        print(f"top_k={top_k} -> {ans[:120]}")
        assert ans
```
