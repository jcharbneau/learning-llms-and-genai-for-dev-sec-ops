## How It Works
1. Preflight checks to make runtime assumptions explicit.
2. A minimal executable example to demonstrate the pattern.
3. Assertions to keep examples test-friendly and reproducible.


```python
try:
    from llama_index.core import Document, VectorStoreIndex
    HAS_LIB = True
except Exception as exc:
    HAS_LIB = False
    ERR = str(exc)

if not HAS_LIB:
    print(f"Skipping: llama-index not installed ({ERR}).")
else:
    docs = [
        Document(text="RAG uses retrieval to ground generated answers."),
        Document(text="Metadata can improve precision in filtered retrieval."),
    ]
    idx = VectorStoreIndex.from_documents(docs)
    qe = idx.as_query_engine(similarity_top_k=1)
    ans = str(qe.query("What is RAG?"))
    print(ans)
    assert "rag" in ans.lower() or "retr" in ans.lower()
```
