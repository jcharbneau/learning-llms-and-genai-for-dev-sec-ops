The next two code cells are intentionally split:

```python
# Deterministic Demo
text = "Security policy requires key rotation. Incident policy defines pager escalation. Cost policy tracks token spend."
small_chunks = [text[i:i+40] for i in range(0, len(text), 40)]
large_chunks = [text[i:i+80] for i in range(0, len(text), 80)]

records = [
    {"section": "security", "chunk": small_chunks[0]},
    {"section": "operations", "chunk": small_chunks[1] if len(small_chunks) > 1 else ""},
    {"section": "finance", "chunk": small_chunks[2] if len(small_chunks) > 2 else ""},
]

filtered = [r for r in records if r["section"] == "security"]
print({"small_count": len(small_chunks), "large_count": len(large_chunks), "filtered": filtered})
assert len(filtered) == 1
```

```python
# Live Demo
try:
    from llama_index.core import Document, VectorStoreIndex
except Exception as exc:
    print(f"Skipping live chunking demo: dependency missing ({exc})")
else:
    docs = [
        Document(text="Security policy mandates key rotation every 90 days.", metadata={"section": "security"}),
        Document(text="Finance policy tracks per-team token budgets.", metadata={"section": "finance"}),
    ]
    idx = VectorStoreIndex.from_documents(docs)
    qe = idx.as_query_engine(similarity_top_k=2)
    ans = str(qe.query("Which policy mentions key rotation?"))
    print(ans)
    assert ans
```
