```python
# Deterministic Demo
rows = [
    {"q": "what grounds answers", "relevant": "D1", "ranked": ["D2", "D1", "D3"]},
    {"q": "how to avoid duplicate side effects", "relevant": "D4", "ranked": ["D4", "D3", "D1"]},
]

hits = 0
rr_total = 0.0
for row in rows:
    rel = row["relevant"]
    ranked = row["ranked"]
    if rel in ranked[:2]:
        hits += 1
    rr_total += 1 / (ranked.index(rel) + 1)

hit_at_2 = hits / len(rows)
mrr = rr_total / len(rows)
print({"hit@2": round(hit_at_2, 2), "mrr": round(mrr, 2)})
assert hit_at_2 >= 0.5
```

```python
# Live Demo
try:
    from llama_index.core import Document, VectorStoreIndex
except Exception as exc:
    print(f"Skipping live retrieval eval demo: dependency missing ({exc})")
else:
    docs = [
        Document(text="RAG grounds responses with retrieved context."),
        Document(text="Idempotency avoids duplicate side effects in retries."),
    ]
    idx = VectorStoreIndex.from_documents(docs)
    nodes = idx.as_retriever(similarity_top_k=2).retrieve("How are responses grounded?")
    texts = [n.text.lower() for n in nodes]
    hit = any("ground" in t for t in texts)
    print("retrieved texts:", texts)
    print("hit:", hit)
    assert isinstance(hit, bool)
```
