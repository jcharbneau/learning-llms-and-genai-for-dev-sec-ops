## Code Walkthrough
The next two code cells are intentionally split:

```python
# Deterministic Demo
def route_query(q: str) -> str:
    ql = q.lower()
    if "policy" in ql or "compliance" in ql:
        return "policy"
    if "python" in ql or "code" in ql or "function" in ql:
        return "code"
    return "general"

queries = [
    "show policy requirements for key rotation",
    "write python retry function",
    "what is our platform overview",
]

routes = {q: route_query(q) for q in queries}
print(routes)
assert routes[queries[0]] == "policy"
```

```python
# Live Demo
try:
    from llama_index.core import Document, VectorStoreIndex
except Exception as exc:
    print(f"Skipping live router demo: dependency missing ({exc})")
else:
    policy_idx = VectorStoreIndex.from_documents([Document(text="Compliance policy requires immutable audit logs.")])
    code_idx = VectorStoreIndex.from_documents([Document(text="Python retry functions should cap attempts and backoff.")])

    q = "How should python retries be implemented safely?"
    chosen = code_idx if "python" in q.lower() else policy_idx
    ans = str(chosen.as_query_engine(similarity_top_k=1).query(q))
    print("routed_to:", "code" if chosen is code_idx else "policy")
    print(ans)
    assert ans
```
