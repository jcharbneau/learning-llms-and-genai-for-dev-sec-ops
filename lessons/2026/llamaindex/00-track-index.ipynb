# 2026 LlamaIndex Lessons

## Track Purpose
RAG indexing and retrieval engineering patterns

## Lesson Roadmap
- `01-rag-minimal-modern.ipynb`: Minimal RAG flow - Build a small index and answer grounded questions.
- `02-chunking-and-metadata-strategy.ipynb`: Chunking and metadata strategy - Compare chunking shapes and metadata filters.
- `03-hybrid-retrieval.ipynb`: Hybrid retrieval - Blend lexical and vector retrieval for better recall.
- `04-reranking-pipeline.ipynb`: Reranking pipeline - Improve top-k quality using rerankers.
- `05-router-retrievers.ipynb`: Router retrievers - Route queries to specialized stores dynamically.
- `06-evaluating-retrieval-quality.ipynb`: Retrieval evaluation - Measure answer grounding and retrieval precision.
- `07-index-persistence-refresh.ipynb`: Persistence and refresh - Persist indices and update them safely over time.
- `08-cost-performance-tuning.ipynb`: Cost and performance tuning - Tune ingestion/query tradeoffs for production workloads.


## How To Use This Track
1. Start with `01` for baseline.
2. Move through reliability/ops/security themes.
3. Capture tradeoffs and adoption decisions per lesson.
