## Code Walkthrough
The next two code cells are intentionally split:
- `Deterministic Demo`: always runnable and concept-focused.
- `Live Demo`: executes a real provider/CLI flow with explicit graceful-skip behavior.


```python
# Deterministic Demo
corpus = [
    "RAG combines retrieval and generation for grounded outputs.",
    "Chunk boundaries influence retrieval precision.",
    "Metadata filters reduce irrelevant context.",
]
query = "Why is retrieval important in RAG?"

scores = []
for doc in corpus:
    overlap = len(set(query.lower().split()) & set(doc.lower().split()))
    scores.append((overlap, doc))

best = sorted(scores, key=lambda x: x[0], reverse=True)[0][1]
print("retrieved:", best)
assert "retrieval" in best.lower() or "rag" in best.lower()
```

```python
# Live Demo
try:
    from llama_index.core import Document, VectorStoreIndex
except Exception as exc:
    print(f"Skipping live LlamaIndex demo: dependency missing ({exc})")
else:
    docs = [
        Document(text="RAG grounds answers with retrieved context."),
        Document(text="Evaluation should track retrieval hit quality."),
    ]
    idx = VectorStoreIndex.from_documents(docs)
    qe = idx.as_query_engine(similarity_top_k=1)
    ans = str(qe.query("What grounds answers in RAG?"))
    print(ans)
    assert ans
```
