# LlamaIndex: Minimal RAG Flow (2026)

Goal: show ingestion, indexing, and retrieval with modern LlamaIndex APIs.


```python
import os

try:
    from llama_index.core import Document, VectorStoreIndex
    HAS_LLAMAINDEX = True
except Exception as exc:
    HAS_LLAMAINDEX = False
    LLAMAINDEX_IMPORT_ERROR = str(exc)

print(f"llamaindex available: {HAS_LLAMAINDEX}")
```

## Build and query a tiny index
This example skips cleanly when LlamaIndex is not installed.


```python
if not HAS_LLAMAINDEX:
    print("Skipping: llama-index is not installed. Install with: pip install llama-index")
    print(f"Import error: {LLAMAINDEX_IMPORT_ERROR}")
else:
    docs = [
        Document(text="DevSecOps integrates security into software delivery."),
        Document(text="RAG combines retrieval with generation for grounded answers."),
    ]
    index = VectorStoreIndex.from_documents(docs)
    engine = index.as_query_engine(similarity_top_k=1)
    response = engine.query("What is RAG?")
    text = str(response)
    print(text)
    assert "retr" in text.lower() or "rag" in text.lower()
```
