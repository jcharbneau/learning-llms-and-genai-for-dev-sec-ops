# Classic LangChain Track (Legacy)

This track preserves legacy notebooks for comparison and historical evaluation.

## Important
- Classic runtime is **optional** and is not built by default.
- Rebuild with `CLASSIC=1 make notebooks-build` before running these lessons.
- In Jupyter, select kernel **Python 3 (classic-langchain)**.


## Classic Lesson Folders
- `developer/`
- `operations/`
- `quality/`
- `security/`


## Quick Start
1. `CLASSIC=1 make notebooks-build`
2. `make notebooks-up`
3. Open this folder and run a `*-classic.ipynb` notebook with kernel `Python 3 (classic-langchain)`.
